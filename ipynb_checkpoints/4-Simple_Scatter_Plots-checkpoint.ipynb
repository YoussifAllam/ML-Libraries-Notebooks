```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('seaborn-whitegrid')
import numpy as np
```

```python
x = np.linspace(0, 10, 30)
y = np.sin(x)

plt.figure(figsize = (7,4))
plt.plot(x, y, 'v', color = 'darkred')
```

```python
plt.figure(figsize=(30,15))
plt.plot(x, y, '-p', color='gray',markersize=16, linewidth=4,markeredgecolor='red',markeredgewidth=5)
plt.ylim(-1.2, 1.2);
```

```python
plt.figure(figsize=(15,7))
plt.scatter(x, y, marker='o');
```

```python
rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)
colors = rng.rand(100)
sizes = 1000 * rng.rand(100)

plt.figure(figsize=(30,15))
plt.scatter(x, y, c=colors, s=sizes, alpha=0.8, cmap='viridis')

plt.colorbar();
```
