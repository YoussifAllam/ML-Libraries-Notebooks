```python
mean = [0, 0]
cov = [[1, 1],
       [1, 2]]
x, y = np.random.multivariate_normal(mean, cov, 10000).T
```

```python
plt.hist2d(x, y, bins=30, cmap='coolwarm')
cb = plt.colorbar()
cb.set_label('counts in bin')
```

```python
plt.hist2d(x, y, bins=30, cmap='Blues')
cb = plt.colorbar()
```

```python
plt.hist2d(x, y, bins=30, cmap='Dark2_r')
cb = plt.colorbar()
cb.set_label('counts in bin')
```

```python
plt.hist2d(x, y, bins=30, cmap='Purples')
cb = plt.colorbar()
cb.set_label('counts in bin')
```
