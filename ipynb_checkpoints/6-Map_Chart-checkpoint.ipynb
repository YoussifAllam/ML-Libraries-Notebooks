```python
import plotly.express as px
import warnings 
warnings.filterwarnings("ignore")  


# Load the data
df = px.data.gapminder().query("year == 2007")
df
```

```python
# Create the choropleth map
fig = px.choropleth(df, locations="iso_alpha",
                    color="lifeExp",
                    hover_name="country",
                    title="Global lifeExp in 2007")

fig.update_layout(
    width=1000,
    height=600
)
fig.show()
```

```python
import pandas as pd

# Simulated earthquake data
data = {
    'Latitude': [38.297, -9.142, 35.6895, 64.9631, -35.6751, 40.7128],
    'Longitude': [-122.2855, -175.2026, 139.6917, -19.0208, -71.543, -74.0060],
    'Magnitude': [5.4, 6.9, 5.1, 4.5, 6.8, 5.2],
    'Location': ['California', 'Tonga', 'Japan', 'Iceland', 'Chile', 'New York']
}

df = pd.DataFrame(data)
df
```

```python
fig = px.scatter_geo(
    df,
    lat='Latitude',
    lon='Longitude',
    size='Magnitude',
    hover_name='Location',
    color='Magnitude',
    projection='natural earth',
    title='Global Earthquake Magnitude Representation',
    size_max=15,
)
fig.update_layout(
    width=1000,
    height=600
)
fig.show()
```

```python
import pandas as pd

data = {
    'City': ['Cairo', 'Alexandria', 'Giza', 'Luxor', 'Aswan', 'Sharm El Sheikh'],
    'Latitude': [30.033333, 31.215645, 30.013056, 25.687243, 24.088938, 27.915817],
    'Longitude': [31.233334, 29.955266, 31.208853, 32.639635, 32.899829, 34.329605]
}

df = pd.DataFrame(data)
```

```python
import plotly.express as px

fig = px.scatter_geo(df,
                     lat='Latitude',
                     lon='Longitude',
                     hover_name='City',
                     projection="natural earth",
                     scope='africa', # Zooms the map to Africa, but Egypt will be the main focus due to plotted points
                     title='Major Cities in Egypt')

# Zoom in closer to Egypt
fig.update_geos(fitbounds="locations", visible=True)
fig.update_layout(
    width=1000,
    height=600
)
fig.show()
```
