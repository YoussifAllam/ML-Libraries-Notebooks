```python
x = np.linspace(0, 20, 1000)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, '-b', label = 'sine')
plt.plot(x, y2, '--r', label = 'cosine')
plt.legend()
plt.ylim(-1.5, 2.0)
```

```python
x1 = np.linspace(0, 10, 20)
y1 = np.sin(x1)

x2 = np.linspace(0, 10, 1000)
y2 = np.sin(x2)
```

```python
plt.plot(x1, y1, '-bo', label = 'sampled')
plt.plot(x2, y2, ':r', label = 'continuous')
plt.legend()

plt.ylim(-1.5, 2.0)
```
