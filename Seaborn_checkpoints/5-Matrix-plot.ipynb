# Tutorial: Creating Heatmaps (Matrix Plots) with Seaborn in Python

Heatmaps, also known as matrix plots, are graphical representations of data where values in a matrix are represented as colors. They are useful for visualizing relationships, patterns, and trends in large datasets. Seaborn, a Python data visualization library built on top of matplotlib, provides a simple and intuitive interface for creating heatmaps.

## 1. Introduction to Heatmaps

Heatmaps are graphical representations of data where values in a matrix are represented as colors. They are commonly used to visualize correlation matrices, frequency tables, and other types of two-dimensional data. In a heatmap, each cell's color intensity is determined by its value, making it easy to identify patterns and relationships within the dataset.

## 3. Loading Sample Data

For this tutorial, we'll use Seaborn's built-in dataset called "flights," which contains information about the number of passengers traveling in different months and years.

````python
import seaborn as sns

# Load dataset (flights)
flights = sns.load_dataset("flights")
flights
````


## 4. Creating a Basic Heatmap

Now, let's create a basic heatmap using the `heatmap()` function from Seaborn.


````python
flights.month.dtype
````

````python
flights.pivot?
````

````python
import matplotlib.pyplot as plt

# Pivot the dataset to create a matrix format

flights_matrix = flights.pivot(index= "month", columns ="year", values ="passengers")
# print(flights_matrix)
# Create a heatmap

flights_matrix
````

````python
plt.figure(figsize=(12,8))
sns.heatmap(flights_matrix, annot=True, cmap="viridis", fmt="d") #cmap --- coolwarm - viridis -- Blues 
plt.title('Passenger Count by Month and Year')
plt.show()
````


In this example, we pivot the dataset to create a matrix format where rows represent months, columns represent years, and the values represent the number of passengers. We then use `sns.heatmap()` to create a heatmap with annotations (`annot=True`), integer formatting (`fmt="d"`), and a specific color map (`cmap="YlGnBu"`).

## 5. Customizing Heatmaps

You can customize heatmaps in various ways, including changing the color map, adjusting annotations, modifying axis labels, and more. Here are some common customizations:

- **Color Map**: Choose from various color maps provided by Matplotlib or create custom color maps.
- **Annotations**: Control whether to display annotations in each cell and format them accordingly.
- **Axes Labels**: Customize the labels for the x-axis and y-axis.
- **Title**: Add a title to the heatmap using `plt.title()`.
- **Figure Size**: Adjust the size of the figure using `plt.figure(figsize=(width, height))`.

````python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load dataset (Iris)
iris = sns.load_dataset("iris")
print(iris[iris.columns[:-1]])
# Calculate correlation matrix
correlation_matrix = iris[iris.columns[:-1]].corr()
````

````python
print(correlation_matrix)
````

````python
# Create a heatmap of correlation matrix
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")
plt.title('Correlation Matrix of Iris Dataset')
plt.show()
````
