```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
tips = sns.load_dataset("tips")
tips 
```

Categorical Plot Types
Seaborn provides various types of plots for visualizing categorical data. Some common ones include:

* Bar Plot: Show point estimates and confidence intervals as rectangular bars.


* Count Plot: Show the count of observations in each category.
* Box Plot: Show the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable.
* Violin Plot: Show the distribution of quantitative data across several levels of one (or more) categorical variables such that those distributions can be compared.
* Strip Plot: Draw a scatter plot where one variable is categorical.
* Swarm Plot: Draw a categorical scatter plot with non-overlapping points.

```python
sns.barplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title('Average Total Bill by Day')
plt.show()
```

```python
tips["day"].value_counts()
```

```python
tips["day"].value_counts().plot(kind="barh")
```

```python
sns.countplot(x="day", data=tips)
plt.title('Count of Observations by Day')
plt.show()
```

```python
sns.countplot(x="day", data=tips, hue="sex")
plt.title('Count of Observations by Day')
plt.show()
```

```python
sns.boxplot(x="day", y="total_bill", data=tips)
plt.title('Distribution of Total Bill by Day')
plt.show()
```

```python
sns.violinplot(x="day", y="total_bill", data=tips, hue="day")
plt.title('Distribution of Total Bill by Day')
plt.show() # A wider section indicates that more observations have that value, while narrower sections indicate fewer observations.
```

```python
sns.stripplot(x="day", y="total_bill", data=tips, hue="day")
plt.title('Total Bill by Day')
plt.show()
```

```python
sns.swarmplot(x="day", y="total_bill", data=tips)
plt.title('Total Bill by Day')
plt.show()
```
