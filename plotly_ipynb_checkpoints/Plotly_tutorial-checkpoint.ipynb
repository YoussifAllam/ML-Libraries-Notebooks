```python
import plotly.express as px 
import pandas as pd 
```

```python
df = px.data.gapminder().query("continent == 'Europe'")

df 
```

```python
fig = px.scatter(df, x="lifeExp" ,y="gdpPercap")
fig.show()
```

```python
fig = px.scatter(df, x="lifeExp" ,y="gdpPercap", size="pop", color="iso_alpha")
fig.show()
```

```python
fig = px.scatter(df, x="lifeExp" ,y="gdpPercap", size="pop", color="lifeExp", hover_name="country")
fig.show()
fig = px.line(df, x="year" , y="lifeExp" , color="country", hover_name="iso_alpha")
fig.show()
fig = px.bar(df, x="country", y="lifeExp", color="country", hover_name="pop", color_continuous_scale="ice")
fig.show()
```
